## Attack class categorizaton

```python
DoS=["back", "land", "neptune", "pod", "smurf", "teardrop", "apache2",  "udpstorm", "processtable", "worm"]
Probe=["satan", "ipsweep", "nmap", "portsweep", "mscan", "saint"]
R2L=["guess_passwd", "ftp_write", "imap", "phf", "multihop", "warezmaster", "warezclient", "spy", "xlock",
           "xsnoop", "snmpguess", "snmpgetattack", "httptunnel", "sendmail", "named"]
U2R=["buffer_overflow","loadmodule","rootkit", "perl","sqlattack", "xterm", "ps"]
Normal=['normal']
```

```python
label_5_dict={'Normal':Normal,'DoS':DoS,'Probe':Probe,'R2L':R2L,'U2R':U2R}
```

```python
label_5_dict
```

### Function for attack class

```python
def attack_class(a):
    for key,value in label_5_dict.items():
           if a in value:
                return key
    return "unknown"
```

```python
kdd_train["label_5"] = kdd_train["attack"].apply(lambda x: attack_class(x))
kdd_test["label_5"] = kdd_test["attack"].apply(lambda x: attack_class(x))
```

```python
kdd_train["label_5"].value_counts()
```

```python
kdd_test["label_5"].value_counts()
```

```python
kdd_test[kdd_test["label_5"]=="unknown"]['attack'].value_counts()
```

```python
kdd_train['label_2']=kdd_train['attack'].apply(lambda x: "normal" if x=='normal' else "attack")
kdd_test['label_2']=kdd_test['attack'].apply(lambda x: "normal" if x=='normal' else "attack")
```
